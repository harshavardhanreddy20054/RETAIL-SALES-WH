```sql
%sql

-- =====================================================
-- FACTSALES VALIDATIONS
-- =====================================================



-- =====================================================
-- VALIDATION 1:
-- Validate total records loaded
-- =====================================================

SELECT
    COUNT(*) AS total_fact_records

FROM `retail-sales`.02_silver.FactSales;



-- =====================================================
-- VALIDATION 2:
-- Validate SalesSK uniqueness
-- Expected Result: 0 rows
-- =====================================================

SELECT
    SalesSK,
    COUNT(*) AS duplicate_count

FROM `retail-sales`.02_silver.FactSales

GROUP BY SalesSK

HAVING COUNT(*) > 1;



-- =====================================================
-- VALIDATION 3:
-- Validate TransactionID uniqueness
-- Expected Result: 0 rows
-- =====================================================

SELECT
    TransactionID,
    COUNT(*) AS duplicate_count

FROM `retail-sales`.02_silver.FactSales

GROUP BY TransactionID

HAVING COUNT(*) > 1;



-- =====================================================
-- VALIDATION 4:
-- Validate no NULL foreign keys
-- Expected Result: 0 rows
-- =====================================================

SELECT
    *

FROM `retail-sales`.02_silver.FactSales

WHERE CustomerSK IS NULL
   OR ProductSK IS NULL
   OR StoreSK IS NULL;



-- =====================================================
-- VALIDATION 5:
-- Validate referential integrity with dimensions
-- Expected Result: 0 rows
-- =====================================================

SELECT
    fs.*

FROM `retail-sales`.02_silver.FactSales fs

LEFT JOIN `retail-sales`.02_silver.DimCustomer dc
    ON fs.CustomerSK = dc.CustomerSK

LEFT JOIN `retail-sales`.02_silver.DimProduct dp
    ON fs.ProductSK = dp.ProductSK

LEFT JOIN `retail-sales`.02_silver.DimStore ds
    ON fs.StoreSK = ds.StoreSK

WHERE dc.CustomerSK IS NULL
   OR dp.ProductSK IS NULL
   OR ds.StoreSK IS NULL;



-- =====================================================
-- VALIDATION 6:
-- Validate Amount calculation
-- Amount = Quantity * UnitPrice
-- Expected Result: 0 rows
-- =====================================================

SELECT
    fs.TransactionID,
    fs.Quantity,
    dp.UnitPrice,
    fs.Amount,
    (fs.Quantity * dp.UnitPrice) AS expected_amount

FROM `retail-sales`.02_silver.FactSales fs

INNER JOIN `retail-sales`.02_silver.DimProduct dp
    ON fs.ProductSK = dp.ProductSK

WHERE fs.Amount != (fs.Quantity * dp.UnitPrice);



-- =====================================================
-- VALIDATION 7:
-- Validate Quantity > 0
-- Expected Result: 0 rows
-- =====================================================

SELECT
    *

FROM `retail-sales`.02_silver.FactSales

WHERE Quantity <= 0;



-- =====================================================
-- VALIDATION 8:
-- Validate TxnDate format and null handling
-- Expected Result: 0 rows
-- =====================================================

SELECT
    *

FROM `retail-sales`.02_silver.FactSales

WHERE TxnDate IS NULL;



-- =====================================================
-- VALIDATION 9:
-- Validate source vs target row count
-- =====================================================

SELECT

    (SELECT COUNT(*)
     FROM `retail-sales`.01_bronze.sales_raw)
        AS source_count,

    (SELECT COUNT(*)
     FROM `retail-sales`.02_silver.FactSales)
        AS target_count;



-- =====================================================
-- VALIDATION 10:
-- Final FactSales summary
-- =====================================================

SELECT

    COUNT(*) AS total_transactions,

    SUM(Quantity) AS total_quantity_sold,

    SUM(Amount) AS total_sales_amount,

    MIN(TxnDate) AS earliest_transaction,

    MAX(TxnDate) AS latest_transaction

FROM `retail-sales`.02_silver.FactSales;
```

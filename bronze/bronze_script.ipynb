```python
%sql
USE CATALOG `retail-sales`;
```

```python
%sql
USE SCHEMA 01_bronze;
```

```python
%sql
DROP TABLE IF EXISTS customers_raw;

CREATE TABLE customers_raw
USING CSV
OPTIONS (
  path 's3://retail-sales-data-wh/raw/customers/',
  header 'true',
  inferSchema 'true' 
);
```

```python
%sql
select * from customers_raw limit 10;
```

```python
%sql

drop table if exists products_raw;
CREATE TABLE products_raw
USING CSV
OPTIONS (
  path 's3://retail-sales-data-wh/raw/products/',
  header 'true',
  inferSchema 'true'
);

drop table if exists stores_raw;
CREATE TABLE stores_raw
USING CSV
OPTIONS (
  path 's3://retail-sales-data-wh/raw/stores/',
  header 'true',
  inferSchema 'true'
);
```

```python
%sql
SELECT * FROM customers_raw LIMIT 10;
SELECT * FROM products_raw LIMIT 10;
SELECT * FROM stores_raw LIMIT 10;
```

```python
%sql
DROP TABLE IF EXISTS sales_raw;

CREATE TABLE sales_raw
USING CSV
OPTIONS (
  path 's3://retail-sales-data-wh/raw/sales/',
  header 'true',
  inferSchema 'true'
);
```

```python
%sql
SELECT * FROM sales_raw LIMIT 10;
```

```python
%sql
-- Verify all 4 bronze tables are accessible
SELECT 'customers' AS tbl, COUNT(*) AS rows FROM `retail-sales`.01_bronze.customers_raw UNION ALL
SELECT 'products',          COUNT(*)           FROM `retail-sales`.01_bronze.products_raw  UNION ALL
SELECT 'sales',             COUNT(*)           FROM `retail-sales`.01_bronze.sales_raw     UNION ALL
SELECT 'stores',            COUNT(*)           FROM `retail-sales`.01_bronze.stores_raw;
```
